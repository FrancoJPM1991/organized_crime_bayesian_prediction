```python
import pandas as pd
import numpy as np
import geopandas as gpd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from PIL import Image
import os
from pyprojroot import here
```

```python
RAW_DATA_DIR = here() / "data" / "raw"
INTERIM_DATA_DIR = here() / "data" / "interim"
PROCESSED_DATA_DIR = here() / "data" / "processed"
RESULTS_DATA_DIR = here() / "results"
```

```python
gdf = gpd.read_file(RAW_DATA_DIR/"geographic/municipalities/00mun.shp")
glocal_results = pd.read_csv(RESULTS_DATA_DIR/"raw_esda_results/yearly_glocal_results.csv", encoding='latin-1')
```

```python
glocal_results["CVEGEO"] = glocal_results["CVEGEO"].astype(str).str.zfill(5)  
gdf["CVEGEO"] = gdf["CVEGEO"].astype(str).str.zfill(5)  
```

```python
QUADRANT_LABELS = {1: "HS", 2: "CS"}

CLUSTER_COLORS = {
    "HS": "#d7191c",   
    "CS": "#2c7bb6",   
    "ns": "#f0f0f0",   
    "island": "#ffffff",
}

years = sorted(glocal_results["year"].unique())
```

```python
frame_paths = []
for year, year_df in glocal_results.groupby("year"):
    plot_gdf = gdf.merge(year_df[["CVEGEO", "cluster"]], on="CVEGEO", how="left")

    fig, ax = plt.subplots(figsize=(10, 10))
    for cluster_type, color in CLUSTER_COLORS.items():
        subset = plot_gdf[plot_gdf["cluster"] == cluster_type]
        subset.plot(ax=ax, color=color, edgecolor="none", linewidth=0)
    plot_gdf.boundary.plot(ax=ax, linewidth=0.1, color="#999999")
    ax.set_title(f"Crime counts G_Local Clusters — {year}", fontsize=16, fontweight="bold")
    ax.axis("off")
    legend_patches = [
        mpatches.Patch(color=CLUSTER_COLORS["HS"], label="Hotspots"),
        mpatches.Patch(color=CLUSTER_COLORS["CS"], label="Coldspots"),
        mpatches.Patch(color=CLUSTER_COLORS["ns"], label="Not significant"),
    ]
    ax.legend(handles=legend_patches, loc="lower left", fontsize=10, frameon=True)
    plt.savefig(RESULTS_DATA_DIR / f"raw_esda_results/glocal_{year}.png", dpi=150, bbox_inches="tight")
    plt.close(fig)
    frame_paths.append(RESULTS_DATA_DIR / f"raw_esda_results/glocal_{year}.png")
    print(f"Saved frame for {year}")
```

```python
frames = [Image.open(p) for p in frame_paths]
max_w = max(f.width for f in frames)
max_h = max(f.height for f in frames)
padded = []
for f in frames:
    bg = Image.new("RGB", (max_w, max_h), "white")
    bg.paste(f, ((max_w - f.width) // 2, (max_h - f.height) // 2))
    padded.append(bg)

gif_path = RESULTS_DATA_DIR/"raw_esda_results/yearly_glocal_clusters.gif"
os.makedirs(RESULTS_DATA_DIR/"raw_esda_results", exist_ok=True)
padded[0].save(
    gif_path,
    save_all=True,
    append_images=padded[1:],
    duration=1000,   # ms per frame
    loop=0,
)
print(f"GIF saved to {gif_path}")
```
